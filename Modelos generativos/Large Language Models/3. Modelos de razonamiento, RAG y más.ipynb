Las LLMs están llenas de acrónimos dado que una vez que disponemos de un modelo fundacional (base) existen multiples formas en las que podemos adaptar o condicionar sus respuestas.

Existe un concepto conocido como **fine-tuning** (afinado) donde tomamos una serie de datos y continuamos ajustando el modelo a este caso particular. Esto implica muchas veces el uso de ajustes de tipología concreta.

**LoRA** (Low-Rank Adaptation) es una técnica utilizada para adaptar modelos de lenguaje de gran tamaño (LLMs) de manera eficiente y económica. En lugar de ajustar todos los parámetros del modelo, LoRA introduce matrices de bajo rango en ciertas capas, permitiendo modificar el comportamiento del modelo con un número reducido de parámetros adicionales. Esto facilita el fine-tuning en tareas específicas sin necesidad de grandes recursos computacionales ni de reentrenar el modelo completo, manteniendo la calidad y flexibilidad en las respuestas generadas.

**QLoRA** (Quantized Low-Rank Adaptation) es una extensión de LoRA que permite realizar el fine-tuning de modelos de lenguaje de gran tamaño utilizando versiones cuantizadas del modelo base. Al aplicar técnicas de cuantización, QLoRA reduce significativamente el uso de memoria y los requisitos computacionales, haciendo posible adaptar LLMs en hardware más limitado, como GPUs de menor capacidad. Esta técnica mantiene la eficiencia y flexibilidad de LoRA, permitiendo entrenar modelos personalizados de manera accesible y rápida, sin sacrificar la calidad de las respuestas generadas.

Los modelos de razonamiento en las LLMs (Large Language Models) representan una de las áreas más fascinantes y prometedoras dentro del campo de la inteligencia artificial. Estos modelos, como GPT-4, han revolucionado la forma en que las máquinas procesan, comprenden y generan lenguaje natural, permitiendo aplicaciones que van desde la asistencia conversacional hasta la resolución de problemas complejos en dominios especializados.

El razonamiento en las LLMs se refiere a la capacidad de estos sistemas para analizar información, establecer relaciones lógicas, inferir conclusiones y tomar decisiones basadas en datos textuales. A diferencia de los enfoques tradicionales de IA, que suelen depender de reglas explícitas o algoritmos simbólicos, las LLMs emplean arquitecturas de redes neuronales profundas entrenadas con enormes cantidades de texto. Esto les permite aprender patrones, contextos y estructuras lingüísticas de manera implícita, lo que resulta en una sorprendente habilidad para simular procesos de razonamiento humano.

Existen diferentes tipos de razonamiento que las LLMs pueden abordar, como el razonamiento deductivo, inductivo y abductivo. El razonamiento deductivo implica extraer conclusiones lógicas a partir de premisas conocidas, mientras que el inductivo se basa en la generalización a partir de ejemplos específicos. El razonamiento abductivo, por su parte, permite generar hipótesis plausibles ante información incompleta. Aunque las LLMs no razonan de manera consciente como los humanos, su entrenamiento masivo les permite aproximarse a estos procesos mediante la generación de respuestas coherentes y contextualmente relevantes.

Sin embargo, los modelos de razonamiento en las LLMs presentan desafíos importantes. Uno de los principales es la tendencia a generar respuestas plausibles pero incorrectas, debido a la naturaleza estadística de su aprendizaje. Además, la interpretación de su razonamiento es compleja, ya que los procesos internos de las redes neuronales profundas no son fácilmente explicables. Por ello, la investigación actual se centra en mejorar la transparencia, la robustez y la capacidad de los modelos para manejar tareas de razonamiento más sofisticadas.

En resumen, los modelos de razonamiento en las LLMs constituyen una herramienta poderosa para la automatización y el apoyo en la toma de decisiones, pero aún requieren avances significativos para alcanzar niveles de comprensión y fiabilidad comparables al razonamiento humano. Su desarrollo continuo promete transformar múltiples sectores, desde la educación y la medicina hasta la ingeniería y la investigación científica.

El aprendizaje por reforzamiento es una rama de la inteligencia artificial en la que un agente aprende a tomar decisiones mediante la interacción con un entorno. El agente ejecuta acciones y recibe recompensas o penalizaciones según el resultado de dichas acciones, ajustando su comportamiento para maximizar la recompensa acumulada a lo largo del tiempo. Este enfoque se utiliza en tareas donde la solución óptima no es evidente y debe descubrirse a través de la experiencia, como juegos, robótica y sistemas de recomendación. El aprendizaje por reforzamiento combina exploración y explotación, permitiendo al agente mejorar progresivamente su desempeño.

````python
import time
import gymnasium as gym

# Create our training environment - a cart with a pole that needs balancing
env = gym.make("CartPole-v1", render_mode="human")

# Reset environment to start a new episode
observation, info = env.reset()
# observation: what the agent can "see" - cart position, velocity, pole angle, etc.
# info: extra debugging information (usually not needed for basic learning)

print(f"Starting observation: {observation}")
# Example output: [ 0.01234567 -0.00987654  0.02345678  0.01456789]
# [cart_position, cart_velocity, pole_angle, pole_angular_velocity]

episode_over = False
total_reward = 0

while not episode_over:
    # Choose an action: 0 = push cart left, 1 = push cart right
    action = env.action_space.sample()  # Random action for now - real agents will be smarter!

    # Take the action and see what happens
    observation, reward, terminated, truncated, info = env.step(action)

    # reward: +1 for each step the pole stays upright
    # terminated: True if pole falls too far (agent failed)
    # truncated: True if we hit the time limit (500 steps)

    total_reward += reward
    episode_over = terminated or truncated

    time.sleep(0.01) # 2 seconds

print(f"Episode finished! Total reward: {total_reward}")
env.close()
````

## Retrieve Augmented Generation

Las LLMs solo pueden basar sus predicciones/generaciones en el conocimiento que han visto/aprendido. Si queremos que se enfoquen en un contenido concreto o incluir información que no fue parte del entrenamiento deberemos disponer un sistema de _búsqueda_ e incluir esa información en nuestro prompt:

* **Retrieve**: Proceso que evalúa la consulta y obtiene información adicional
* **Augmented**: Incluimos el contexto informado en el prompt final
* **Generation**: Generamos una respuesta en base a lo consultado y la información disponible

## Model Context Protocol (MCP)

El Model Context Protocol (MCP) es un enfoque diseñado para mejorar la interacción entre modelos de lenguaje y sistemas externos, facilitando el intercambio eficiente de información contextual relevante durante el proceso de generación de respuestas. MCP define una estructura estandarizada para comunicar el estado, las restricciones y el contexto de la tarea al modelo, permitiendo que las respuestas sean más precisas y adaptadas a las necesidades específicas del usuario o aplicación. Este protocolo es especialmente útil en escenarios donde el modelo debe integrar datos dinámicos, instrucciones específicas o información recuperada de fuentes externas, optimizando así la utilidad y la fiabilidad de las soluciones generadas por sistemas de inteligencia artificial.

````python
import random
from pydantic_ai import Agent, RunContext

agent = Agent(
    'google-gla:gemini-1.5-flash',  
    deps_type=str,  
    system_prompt=(
        "Eres un jugador de dados. Tiras el dado y ves el número que sale"
        "Recibes respuestas de si el usuario acertó. Si es que sí, el usuario gana."
        "Usa el nombre del usuario al felicitarle."
    ),
)


@agent.tool_plain  
def roll_dice() -> str:
    """Tiramos el dado"""
    return str(random.randint(1, 6))


@agent.tool  
def get_player_name(ctx: RunContext[str]) -> str:
    """Preguntamos por su nombre"""
    return ctx.deps
````

````python
result = agent.run_sync("Me llamo Iraitz y creo que va a salir el 6")
print(result.output)
````
